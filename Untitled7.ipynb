```python
import sys
sys.path.insert(0, r'C:\Users\Bo\projects\neuromorphic-sandbox')
from src.topology import TopologyGenerator
import numpy as np

print("=" * 60)
print("EXPERIMENT 4: HEAT-CONSTRAINED PLACEMENT")
print("Finding optimal cluster tightness under thermal limits")
print("=" * 60)

def build_k_nearest(coords, k=10):
    n = len(coords)
    adjacency = np.zeros((n, n))
    for i in range(n):
        dists = np.linalg.norm(coords - coords[i], axis=1)
        dists[i] = np.inf
        nearest = np.argsort(dists)[:k]
        adjacency[i, nearest] = 1
    return adjacency

def comm_energy(coords, adjacency):
    """Communication energy: sum of distance^2"""
    total = 0
    for i in range(len(coords)):
        for j in np.where(adjacency[i] > 0)[0]:
            total += np.linalg.norm(coords[i] - coords[j]) ** 2
    return total

def thermal_penalty(coords, cell_size=1.0, density_limit=8, penalty_scale=50):
    """Divide space into cells. Neurons per cell above the limit
    incur a superlinear (quadratic) penalty - modeling heat buildup."""
    # Bin neurons into grid cells
    mins = coords.min(axis=0)
    cells = {}
    for c in coords:
        key = tuple(((c - mins) // cell_size).astype(int))
        cells[key] = cells.get(key, 0) + 1
    
    penalty = 0
    max_density = 0
    for count in cells.values():
        max_density = max(max_density, count)
        if count > density_limit:
            excess = count - density_limit
            penalty += penalty_scale * excess ** 2  # quadratic: heat compounds
    return penalty, max_density

print("\nSweeping cluster radius from very tight to very loose")
print(f"(5 clusters, 1000 neurons, k=10 connections each)")
print(f"(Thermal limit: 8 neurons per unit cell, quadratic penalty above)\n")

radii = [0.3, 0.5, 0.75, 1.0, 1.5, 2.0, 3.0, 4.0, 6.0]
results = []

for r in radii:
    np.random.seed(42)  # same cluster centers each run - fair comparison
    tg = TopologyGenerator(1000)
    tg.random_cluster(5, cluster_radius=r)
    coords = tg.get_coordinates()
    adj = build_k_nearest(coords, k=10)
    
    e_comm = comm_energy(coords, adj)
    e_heat, max_d = thermal_penalty(coords)
    e_total = e_comm + e_heat
    
    results.append((r, e_comm, e_heat, e_total, max_d))
    print(f"radius={r:4.2f}:  comm={e_comm:10,.0f}  heat={e_heat:12,.0f}  TOTAL={e_total:12,.0f}  (max density: {max_d}/cell)")

print("\n" + "=" * 60)
best = min(results, key=lambda x: x[3])
print(f"OPTIMAL CLUSTER RADIUS: {best[0]}")
print(f"  Communication energy: {best[1]:,.0f}")
print(f"  Thermal penalty:      {best[2]:,.0f}")
print(f"  Total cost:           {best[3]:,.0f}")
print("=" * 60)
print("\nThe curve should be U-shaped:")
print("  Too tight -> thermal penalty explodes")
print("  Too loose -> communication energy explodes")
print("  The optimum is the chip designer's sweet spot.")
```

```python
import sys
sys.path.insert(0, r'C:\Users\Bo\projects\neuromorphic-sandbox')
import numpy as np
from collections import deque

print("EXPERIMENT 7: TRAFFIC & CONGESTION")
print("Routing 2000 random messages through each backbone")
print("=" * 64)

def spread_clusters(n_neurons, n_clusters, radius, center_spread, seed=42):
    coords, cluster_ids = [], []
    per_cluster = n_neurons // n_clusters
    rng = np.random.default_rng(seed)
    centers = rng.uniform(0, center_spread, size=(n_clusters, 2))
    for cid, (cx, cy) in enumerate(centers):
        for _ in range(per_cluster):
            theta = rng.uniform(0, 2*np.pi)
            r = rng.uniform(0, radius)
            coords.append([cx + r*np.cos(theta), cy + r*np.sin(theta), 0.0])
            cluster_ids.append(cid)
    return np.array(coords), np.array(cluster_ids), centers

def build_k_nearest(coords, k=10):
    n = len(coords)
    adjacency = np.zeros((n, n))
    for i in range(n):
        dists = np.linalg.norm(coords - coords[i], axis=1)
        dists[i] = np.inf
        adjacency[i, np.argsort(dists)[:k]] = 1
    return adjacency

def find_routers(coords, cluster_ids, centers):
    routers = []
    for cid in range(len(centers)):
        members = np.where(cluster_ids == cid)[0]
        dists = np.linalg.norm(coords[members, :2] - centers[cid], axis=1)
        routers.append(members[np.argmin(dists)])
    return routers

def add_backbone(adjacency, coords, routers, pattern):
    adj = adjacency.copy()
    R = len(routers)
    if pattern == 'mesh':
        links = [(a, b) for a in range(R) for b in range(a+1, R)]
    elif pattern == 'ring':
        links = [(a, (a+1) % R) for a in range(R)]
    elif pattern == 'star':
        links = [(0, b) for b in range(1, R)]
    for a, b in links:
        i, j = routers[a], routers[b]
        adj[i, j] = adj[j, i] = 1
    return adj

def route_traffic(adjacency, n_messages=2000, seed=11):
    """Send messages between random pairs via BFS shortest path.
    Count how many messages transit each node."""
    n = len(adjacency)
    neighbors = [np.where(adjacency[i] > 0)[0] for i in range(n)]
    rng = np.random.default_rng(seed)
    node_load = np.zeros(n)
    total_hops, delivered = 0, 0
    
    for _ in range(n_messages):
        src, dst = rng.integers(0, n, size=2)
        if src == dst:
            continue
        # BFS with parent tracking to reconstruct path
        parent = -2 * np.ones(n, dtype=int)
        parent[src] = -1
        q = deque([src])
        while q and parent[dst] == -2:
            u = q.popleft()
            for v in neighbors[u]:
                if parent[v] == -2:
                    parent[v] = u
                    q.append(v)
        if parent[dst] == -2:
            continue  # unreachable
        # Walk back the path, incrementing load on every transit node
        node = dst
        path_len = 0
        while node != src:
            node_load[node] += 1
            node = parent[node]
            path_len += 1
        node_load[src] += 1
        total_hops += path_len
        delivered += 1
    
    return node_load, total_hops / max(delivered, 1), delivered

# Build everything
coords, cluster_ids, centers = spread_clusters(1000, 5, radius=8, center_spread=40)
base_adj = build_k_nearest(coords, k=10)
routers = find_routers(coords, cluster_ids, centers)

print(f"\nRouters: {routers}")
print(f"\n{'topology':>8} | {'delivered':>9} | {'avg hops':>8} | {'max load':>8} | {'hotspot node':>12} | {'load ratio':>10}")
print("-" * 72)

for pattern in ['star', 'ring', 'mesh']:
    adj = add_backbone(base_adj, coords, routers, pattern)
    load, hops, delivered = route_traffic(adj)
    max_load = load.max()
    hotspot = int(np.argmax(load))
    is_router = "ROUTER" if hotspot in routers else "regular"
    mean_load = load[load > 0].mean()
    ratio = max_load / mean_load
    print(f"{pattern:>8} | {delivered:>9} | {hops:>8.2f} | {max_load:>8.0f} | {hotspot:>5} ({is_router}) | {ratio:>9.1f}x")

print("\n'load ratio' = busiest node vs average active node")
print("A ratio of 1x would be perfectly balanced; high ratios = hotspot")
print("\nRemember Exp 4: concentrated activity = concentrated HEAT.")
print("A congested router is a thermal hotspot wearing a network hat.")
```

```python
import sys
sys.path.insert(0, r'C:\Users\Bo\projects\neuromorphic-sandbox')
import numpy as np
from collections import deque

print("EXPERIMENT 9: 3D STACKING")
print("Folding the 2D chip into vertical layers")
print("=" * 64)

def build_3d_chip(n_neurons, n_clusters, n_layers, base_footprint=40.0,
                  cluster_radius=8.0, layer_gap=1.0, seed=42):
    """Same 5-cluster chip, folded into n_layers tiers.
    Footprint shrinks as sqrt(1/n_layers) to keep total area constant."""
    footprint = base_footprint / np.sqrt(n_layers)
    radius = cluster_radius / np.sqrt(n_layers)
    rng = np.random.default_rng(seed)
    centers = rng.uniform(0, footprint, size=(n_clusters, 2))
    per_cluster = n_neurons // n_clusters
    coords, cluster_ids = [], []
    for cid, (cx, cy) in enumerate(centers):
        for m in range(per_cluster):
            theta = rng.uniform(0, 2*np.pi)
            r = rng.uniform(0, radius)
            z = (m % n_layers) * layer_gap   # deal members across layers
            coords.append([cx + r*np.cos(theta), cy + r*np.sin(theta), z])
            cluster_ids.append(cid)
    return np.array(coords), np.array(cluster_ids)

def wire_energy_3d(ci, cj, tsv_multiplier=3.0):
    """Horizontal distance^2 + penalized vertical distance^2"""
    dxy = np.linalg.norm(ci[:2] - cj[:2])
    dz = abs(ci[2] - cj[2])
    return dxy**2 + tsv_multiplier * dz**2

def build_k_nearest_3d(coords, k=10, tsv_multiplier=3.0):
    """Connect to k neighbors cheapest by 3D wire cost (not raw distance)"""
    n = len(coords)
    adjacency = np.zeros((n, n))
    total_energy = 0
    for i in range(n):
        costs = np.array([wire_energy_3d(coords[i], coords[j], tsv_multiplier)
                          if j != i else np.inf for j in range(n)])
        nearest = np.argsort(costs)[:k]
        adjacency[i, nearest] = 1
        total_energy += costs[nearest].sum()
    return adjacency, total_energy

def thermal_3d(coords, n_layers, cell_size=1.0, density_limit=8,
               penalty_scale=50, layer_gap=1.0):
    """2D density penalty per layer, multiplied by depth factor:
    buried layers dissipate heat worse. Layer L multiplier = 1 + 0.5*L
    (top layer touches heatsink: L=0)"""
    penalty = 0
    max_cell = 0
    for L in range(n_layers):
        layer_mask = np.isclose(coords[:, 2], L * layer_gap)
        layer_pts = coords[layer_mask]
        if len(layer_pts) == 0:
            continue
        depth_mult = 1 + 0.5 * (n_layers - 1 - L)  # bottom = most buried
        mins = layer_pts[:, :2].min(axis=0)
        cells = {}
        for c in layer_pts:
            key = tuple(((c[:2] - mins) // cell_size).astype(int))
            cells[key] = cells.get(key, 0) + 1
        for count in cells.values():
            max_cell = max(max_cell, count)
            if count > density_limit:
                penalty += depth_mult * penalty_scale * (count - density_limit) ** 2
    return penalty, max_cell

def avg_hops(adjacency, n_samples=30, seed=7):
    n = len(adjacency)
    neighbors = [np.where(adjacency[i] > 0)[0] for i in range(n)]
    rng = np.random.default_rng(seed)
    total_hops, total_pairs, unreachable = 0, 0, 0
    for s in rng.integers(0, n, size=n_samples):
        dist = -np.ones(n, dtype=int); dist[s] = 0
        q = deque([s])
        while q:
            u = q.popleft()
            for v in neighbors[u]:
                if dist[v] < 0:
                    dist[v] = dist[u] + 1
                    q.append(v)
        reached = dist[dist > 0]
        total_hops += reached.sum()
        total_pairs += len(reached)
        unreachable += (dist < 0).sum()
    return total_hops / max(total_pairs, 1), unreachable / n_samples

print(f"\n{'layers':>6} | {'footprint':>9} | {'wire energy':>11} | {'thermal':>9} | {'TOTAL':>9} | {'hops':>6} | {'unreach':>7}")
print("-" * 74)

results = []
for n_layers in [1, 2, 4, 8]:
    coords, cluster_ids = build_3d_chip(1000, 5, n_layers)
    adj, wire_e = build_k_nearest_3d(coords, k=10)
    heat_e, max_cell = thermal_3d(coords, n_layers)
    total = wire_e + heat_e
    hops, unreach = avg_hops(adj)
    footprint = 40.0 / np.sqrt(n_layers)
    results.append((n_layers, total))
    print(f"{n_layers:>6} | {footprint:>7.1f}mm | {wire_e:>11,.0f} | {heat_e:>9,.0f} | {total:>9,.0f} | {hops:>6.2f} | {unreach:>7.0f}")

best = min(results, key=lambda x: x[1])
print(f"\nOPTIMAL LAYER COUNT: {best[0]}")
print("\nThe brain's version: cortex is a folded 2D sheet (~2-4mm thick),")
print("not a solid 3D volume. Evolution also stopped at 'a few layers.'")
print("Watch whether the optimum lands there too.")
```

```python
import numpy as np
from collections import deque

print("EXPERIMENT 9B: 3D STACKING WITH SHARED-HEATSINK THERMAL MODEL")
print("Heat must exit through the top: penalize COLUMN density")
print("=" * 64)

# (reuse build_3d_chip, wire_energy_3d, build_k_nearest_3d, avg_hops
#  from the previous cell - they're still defined in the kernel)

def thermal_3d_column(coords, n_layers, cell_size=1.0, density_limit=8,
                      penalty_scale=50):
    """Shared heatsink: sum ALL neurons in each (x,y) column across layers.
    Deeper stacks also insulate: penalty multiplier = 1 + 0.4*(n_layers-1)"""
    mins = coords[:, :2].min(axis=0)
    columns = {}
    for c in coords:
        key = tuple(((c[:2] - mins) // cell_size).astype(int))
        columns[key] = columns.get(key, 0) + 1
    stack_mult = 1 + 0.4 * (n_layers - 1)
    penalty = 0
    max_col = 0
    for count in columns.values():
        max_col = max(max_col, count)
        if count > density_limit:
            penalty += stack_mult * penalty_scale * (count - density_limit) ** 2
    return penalty, max_col

print(f"\n{'layers':>6} | {'footprint':>9} | {'wire energy':>11} | {'thermal':>10} | {'TOTAL':>10} | {'max col':>7}")
print("-" * 68)

results = []
for n_layers in [1, 2, 4, 8]:
    coords, cluster_ids = build_3d_chip(1000, 5, n_layers)
    adj, wire_e = build_k_nearest_3d(coords, k=10)
    heat_e, max_col = thermal_3d_column(coords, n_layers)
    total = wire_e + heat_e
    footprint = 40.0 / np.sqrt(n_layers)
    results.append((n_layers, total))
    print(f"{n_layers:>6} | {footprint:>7.1f}mm | {wire_e:>11,.0f} | {heat_e:>10,.0f} | {total:>10,.0f} | {max_col:>7}")

best = min(results, key=lambda x: x[1])
print(f"\nOPTIMAL LAYER COUNT (honest thermal): {best[0]}")
print("\nPrediction: shallow stacking (2ish) wins now. Column density grows")
print("with layers (smaller footprint + everything piled vertically), and")
print("the stack multiplier compounds it. Deep stacks should get punished.")
```

```python
import sys
sys.path.insert(0, r'C:\Users\Bo\projects\neuromorphic-sandbox')
import numpy as np

print("EXPERIMENT 10: HETEROGENEOUS STACKING (memory-on-logic)")
print("Can stacking win when the stacked layer runs cool?")
print("=" * 64)

TSV_MULT = 3.0
MEM_POWER = 0.1   # memory node power vs compute node
N = 1000

def make_compute(seed=42, footprint=40.0, radius=8.0):
    rng = np.random.default_rng(seed)
    centers = rng.uniform(0, footprint, size=(5, 2))
    coords = []
    for cx, cy in centers:
        for _ in range(N // 5):
            theta = rng.uniform(0, 2*np.pi)
            r = rng.uniform(0, radius)
            coords.append([cx + r*np.cos(theta), cy + r*np.sin(theta), 0.0])
    return np.array(coords)

def memory_access_energy(compute_xy, memory_xy):
    """Each compute node accesses ITS paired memory node.
    Energy = horizontal dist^2 + 3x vertical dist^2, summed over all pairs."""
    total = 0
    for c, m in zip(compute_xy, memory_xy):
        dxy = np.linalg.norm(c[:2] - m[:2])
        dz = abs(c[2] - m[2])
        total += dxy**2 + TSV_MULT * dz**2
    return total

def thermal_power_columns(coords, powers, cell_size=1.0, power_limit=8.0,
                          penalty_scale=50, n_layers=1):
    """Column model from 9B, but sums POWER not headcount"""
    mins = coords[:, :2].min(axis=0)
    columns = {}
    for c, p in zip(coords, powers):
        key = tuple(((c[:2] - mins) // cell_size).astype(int))
        columns[key] = columns.get(key, 0) + p
    stack_mult = 1 + 0.4 * (n_layers - 1)
    penalty = 0
    for pw in columns.values():
        if pw > power_limit:
            penalty += stack_mult * penalty_scale * (pw - power_limit) ** 2
    return penalty

compute = make_compute()
rng = np.random.default_rng(9)

configs = {}

# --- Config 1: memory BESIDE (own region, offset in x) ---
mem_beside = compute.copy()
mem_beside[:, 0] += 45.0   # memory bank sits next to the die
configs['beside (2D)'] = (mem_beside, 1)

# --- Config 2: memory INTERLEAVED (jittered next to each compute node, same plane) ---
mem_inter = compute.copy()
mem_inter[:, :2] += rng.uniform(0.3, 0.8, size=(N, 2))  # crammed alongside
configs['interleaved (2D)'] = (mem_inter, 1)

# --- Config 3: memory STACKED (directly above, one layer up) ---
mem_stack = compute.copy()
mem_stack[:, 2] = 1.0      # same x,y — one vertical hop
configs['stacked (3D)'] = (mem_stack, 2)

print(f"\n{'config':>17} | {'mem access':>10} | {'thermal':>9} | {'TOTAL':>10}")
print("-" * 56)

for name, (mem_coords, n_layers) in configs.items():
    access_e = memory_access_energy(compute, mem_coords)
    all_coords = np.vstack([compute, mem_coords])
    all_powers = np.concatenate([np.ones(N), MEM_POWER * np.ones(N)])
    heat_e = thermal_power_columns(all_coords, all_powers, n_layers=n_layers)
    total = access_e + heat_e
    print(f"{name:>17} | {access_e:>10,.0f} | {heat_e:>9,.0f} | {total:>10,.0f}")

print("\nSame honest thermal model that killed logic-on-logic stacking.")
print("If 'stacked' wins now, we've reproduced WHY HBM exists:")
print("the stacked layer only costs 0.1 power per column — nearly free heat —")
print("while every memory access shrinks from a cross-die trip to one TSV hop.")
```

```python
import sys
sys.path.insert(0, r'C:\Users\Bo\projects\neuromorphic-sandbox')
import numpy as np

print("EXPERIMENT 18a: THE INSTRUMENT — PAIR-CORRELATION FUNCTION ξ(r)")
print("Reading the architecture of our past universes")
print("=" * 66)

def xi_of_r(coords, n_bins=20, r_max=None, n_random=None, seed=99):
    """Cosmology's two-point correlation function via the Peebles-Hauser
    estimator: xi(r) = DD(r)/RR(r) - 1, where DD counts data-data pairs
    at separation r and RR counts pairs in a random catalog of the same
    density over the same footprint."""
    n = len(coords)
    if n_random is None: n_random = n
    lo, hi = coords.min(0), coords.max(0)
    if r_max is None:
        r_max = 0.5 * np.linalg.norm(hi - lo)
    rng = np.random.default_rng(seed)
    randoms = rng.uniform(lo, hi, size=(n_random, 2))

    def pair_dists(pts):
        d = np.sqrt(((pts[:, None, :] - pts[None, :, :])**2).sum(-1))
        return d[np.triu_indices(len(pts), k=1)]

    bins = np.linspace(0, r_max, n_bins + 1)
    DD, _ = np.histogram(pair_dists(coords), bins=bins)
    RR, _ = np.histogram(pair_dists(randoms), bins=bins)
    with np.errstate(divide='ignore', invalid='ignore'):
        xi = np.where(RR > 0, DD / np.maximum(RR, 1) - 1.0, 0.0)
    centers = 0.5 * (bins[1:] + bins[:-1])
    return centers, xi

# ---------- Rebuild historic geometries (exact seeds) ----------
def geom_random_gas(seed=3):
    return np.random.default_rng(seed).uniform(0, 40, size=(1000, 2))

def geom_designed(seed=42):        # Exp 3-14 style clustered die
    rng = np.random.default_rng(seed)
    centers = rng.uniform(0, 40, size=(5, 2))
    pts = []
    for cx, cy in centers:
        for _ in range(200):
            th, r = rng.uniform(0, 2*np.pi), rng.uniform(0, 8)
            pts.append([cx + r*np.cos(th), cy + r*np.sin(th)])
    return np.array(pts)

def rerun_migration(crowd, repel, step_sz, steps=800):
    """Compressed rerun of Exp 15/15B physics to regenerate final coords"""
    N, NC = 1000, 5
    PATTERNS = [(0,3),(1,4),(2,)]
    centers = np.array([[10,10],[30,10],[20,34],[30,16],[10,16]], dtype=float)
    rng = np.random.default_rng(1)
    coords, cids = [], []
    for cid, (cx, cy) in enumerate(centers):
        for _ in range(200):
            th, r = rng.uniform(0, 2*np.pi), rng.uniform(0, 5)
            coords.append([cx + r*np.cos(th), cy + r*np.sin(th)])
            cids.append(cid)
    coords = np.array(coords); cids = np.array(cids)
    rng2 = np.random.default_rng(7)
    src = rng2.integers(0, N, 10000); dst = rng2.integers(0, N, 10000)
    keep = src != dst; src, dst = src[keep], dst[keep]
    inhib = rng2.random(N) < 0.20
    def build_W(s, d):
        W = np.zeros((N, N))
        np.add.at(W, (s, d), np.where(inhib[s], -0.60, 0.30))
        return W
    W = build_W(src, dst)
    v = np.zeros(N); refrac = np.zeros(N, dtype=int)
    C = np.zeros((N, N))
    rng3 = np.random.default_rng(0)
    for t in range(steps):
        pat = PATTERNS[(t // 20) % 3]
        inputs = rng3.uniform(0, 0.02, N)
        if (t % 20) < 5:
            for cid in pat:
                inputs[cids == cid] += 0.5
        v = v * 0.90 + inputs
        fired = (v >= 1.0) & (refrac == 0)
        if fired.any():
            v += W.T @ fired.astype(float)
            f = np.where(fired)[0]
            C[np.ix_(f, f)] += 1.0
        v = np.maximum(v, 0); v[fired] = 0
        refrac[fired] = 3; refrac[refrac > 0] -= 1
        C *= 0.95
        if (t + 1) % 40 == 0:
            Csym = C + C.T
            row = Csym.sum(axis=1, keepdims=True)
            active = row[:, 0] > 1e-6
            targets = np.where(active[:, None],
                               (Csym @ coords) / np.maximum(row, 1e-9), coords)
            pull = targets - coords
            nrm = np.linalg.norm(pull, axis=1, keepdims=True)
            coords = coords + pull * np.minimum(1.0, step_sz / np.maximum(nrm, 1e-9))
            keys = np.floor(coords).astype(int)
            cellmap = {}
            for i, k in enumerate(map(tuple, keys)):
                cellmap.setdefault(k, []).append(i)
            for k, members in cellmap.items():
                if len(members) > crowd:
                    m = np.array(members)
                    cc = coords[m].mean(axis=0)
                    away = coords[m] - cc
                    nn = np.linalg.norm(away, axis=1, keepdims=True)
                    coords[m] += repel * away / np.maximum(nn, 1e-9)
            D2 = ((coords[:, None, :] - coords[None, :, :])**2).sum(-1)
            cold = np.argsort(C[src, dst])[:500]
            mask = C / (1.0 + 0.05 * D2)
            mask[src, dst] = -1
            np.fill_diagonal(mask, -1)
            flat = np.argpartition(mask.ravel(), -500)[-500:]
            ns, nd = np.unravel_index(flat, (N, N))
            src[cold], dst[cold] = ns, nd
            W = build_W(src, dst)
    return coords

print("\nRegenerating universes (blob ~30s, glass ~30s)...")
universes = {
    'random gas (null)': geom_random_gas(),
    'designed die (Exp 3-14)': geom_designed(),
    'BLOB (Exp 15)': rerun_migration(crowd=8, repel=0.4, step_sz=0.6),
    'GLASS (Exp 15B)': rerun_migration(crowd=4, repel=1.2, step_sz=0.4),
}

print("\nξ(r) fingerprints  (each row: separation r | correlation excess)")
print("  ξ>0 clustered at this scale · ξ≈0 random · ξ<0 void/depleted\n")

for name, pts in universes.items():
    r, xi = xi_of_r(pts, n_bins=16, r_max=24)
    print(f"--- {name} ---")
    xmax = max(abs(xi).max(), 1e-9)
    for ri, xv in zip(r, xi):
        bar_len = int(24 * min(abs(xv) / xmax, 1.0))
        bar = ("+" if xv > 0 else "-") * bar_len
        print(f"  r={ri:5.1f} | xi={xv:+7.2f} | {bar}")
    # summary stats
    peak_r = r[np.argmax(xi)]
    zero_cross = r[np.argmax(xi < 0)] if (xi < 0).any() else None
    print(f"  >> clustering peak at r≈{peak_r:.1f}; "
          f"{'void onset r≈%.1f' % zero_cross if zero_cross else 'no void signature'}\n")

print("=" * 66)
print("How to read the fingerprints:")
print("  random gas:  flat near zero everywhere — no structure, the null")
print("  designed:    ξ>0 hump at small r (clusters), ξ<0 trough at mid-r")
print("               (the space BETWEEN clusters) — filament-void class")
print("  blob:        enormous ξ at tiny r, deep negative everywhere else —")
print("               one collapsed object in an empty universe")
print("  glass:       between designed and blob — did frustration preserve")
print("               the two-scale structure or smear it?")
print("\nNext: this instrument goes live inside Exp 18b's constants sweep,")
print("classifying each universe's phase automatically from its ξ curve.")
```

```python
import sys
sys.path.insert(0, r'C:\Users\Bo\projects\neuromorphic-sandbox')
import numpy as np

print("EXPERIMENT 19: THE UTILITY DIAL")
print("Same correlation, different value. Does the network learn what MATTERS?")
print("=" * 70)

N, NC = 1000, 5
PATTERNS = [(0,3), (1,4), (2,)]
REWARDS  = {0: +1.0, 1: 0.0, 2: -1.0}     # rewarded / neutral / punished
STEPS, EPOCH, SWAP = 600, 40, 500

def make_salt_clustered():
    rng = np.random.default_rng(11)
    centers = np.random.default_rng(42).uniform(5, 35, size=(5, 2))
    pts = []
    for cx, cy in centers:
        for _ in range(200):
            th, r = rng.uniform(0, 2*np.pi), rng.uniform(0, 8)
            pts.append([cx + r*np.cos(th), cy + r*np.sin(th)])
    coords = np.array(pts)
    cids = np.repeat(np.arange(NC), 200)
    rng.shuffle(cids)                       # scatter the roles
    return coords, cids

def live(rule, seed_life=0):
    coords, cids = make_salt_clustered()
    D2 = ((coords[:,None,:]-coords[None,:,:])**2).sum(-1)
    rng2 = np.random.default_rng(7)
    src = rng2.integers(0, N, 10000); dst = rng2.integers(0, N, 10000)
    keep = src != dst; src, dst = src[keep], dst[keep]
    inhib = rng2.random(N) < 0.20
    def bW(s, d):
        W = np.zeros((N, N))
        np.add.at(W, (s, d), np.where(inhib[s], -0.60, 0.30))
        return W
    W = bW(src, dst)
    v = np.zeros(N); refrac = np.zeros(N, dtype=int)
    C = np.zeros((N, N))      # slow correlation (the old god)
    E = np.zeros((N, N))      # fast eligibility trace
    V = np.zeros((N, N))      # value (the new god)
    rng3 = np.random.default_rng(seed_life)
    log = []
    for t in range(STEPS):
        p = (t // 20) % 3
        pat = PATTERNS[p]
        inputs = rng3.uniform(0, 0.02, N)
        if (t % 20) < 5:
            for cid in pat: inputs[cids == cid] += 0.5
        v_ = v*0.90 + inputs
        fired = (v_ >= 1.0) & (refrac == 0)
        if fired.any():
            v_ += W.T @ fired.astype(float)
            f = np.where(fired)[0]
            C[np.ix_(f, f)] += 1.0
            E[np.ix_(f, f)] += 1.0
        v = np.maximum(v_, 0); v[fired] = 0
        refrac[fired] = 3; refrac[refrac > 0] -= 1
        C *= 0.95
        E *= 0.90                                  # eligibility decays fast
        if (t % 20) == 6:                          # reward broadcast, 1 step late
            R = REWARDS[p]
            if R != 0.0:
                V += R * E
        V *= 0.999                                 # value barely decays

        if (t+1) % EPOCH == 0:
            score = V if rule == 'utility' else C
            cold = np.argsort(score[src, dst])[:SWAP]
            if rule == 'utility':
                mask = (np.maximum(V, 0) + 0.01*C) / (1.0 + 0.05*D2)
            else:
                mask = C / (1.0 + 0.05*D2)
            mask[src, dst] = -1; np.fill_diagonal(mask, -1)
            flat = np.argpartition(mask.ravel(), -SWAP)[-SWAP:]
            ns, nd = np.unravel_index(flat, (N, N))
            src[cold], dst[cold] = ns, nd
            W = bW(src, dst)
            M = np.zeros((NC, NC), dtype=int)
            np.add.at(M, (cids[src], cids[dst]), 1)
            RB = M[0,3]+M[3,0]          # rewarded bridge
            UB = M[1,4]+M[4,1]          # unrewarded bridge
            BASE = M[0,1]+M[1,0]        # never-co-active chance reference
            P2 = M[2,2]                 # punished cluster's internal wiring
            log.append((t+1, RB, UB, BASE, P2, D2[src,dst].sum()))
    return log, M

print("\nLiving under UTILITY rule (retain what paid off)...")
logU, MU = live('utility')
print("Living under CORRELATION rule (retain what co-fired) — control...")
logC, MC = live('correlation')

def show(log, title):
    print(f"\n{title}")
    print(f"{'step':>5} | {'rewarded':>8} | {'neutral':>7} | {'chance':>6} | {'punished-int':>12} | {'energy':>10}")
    print(f"      |  (0,3)+1  | (1,4) 0 | (0,1)  |   C2 self -1  |")
    print("-" * 70)
    for t, RB, UB, BASE, P2, e in log:
        print(f"{t:>5} | {RB:>8} | {UB:>7} | {BASE:>6} | {P2:>12} | {e:>10,.0f}")

show(logU, "UTILITY RULE:")
show(logC, "CORRELATION RULE (control):")

print("\n" + "=" * 70)
u = logU[-1]; c = logC[-1]
print(f"Selectivity (rewarded/neutral):  utility {u[1]/max(u[2],1):.2f}x  |  "
      f"correlation {c[1]/max(c[2],1):.2f}x")
print(f"Punished-cluster internal wiring: utility {u[4]}  |  correlation {c[4]}")
print("\nThe test: correlation can't tell (0,3) from (1,4) — same statistics.")
print("If the utility network builds ONLY the valuable bridge, suppresses")
print("the punished structure, and the control builds both equally — then")
print("retention has been captured by usefulness. Galaxy -> organism.")
```
