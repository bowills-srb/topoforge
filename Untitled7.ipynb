```python
import sys
sys.path.insert(0, r'C:\Users\Bo\projects\neuromorphic-sandbox')
from src.topology import TopologyGenerator
import numpy as np

print("=" * 60)
print("EXPERIMENT 3: NORMALIZED COMPARISON")
print("Same connection count (10,000) for every topology")
print("=" * 60)

def build_k_nearest(coords, k=10):
    """Connect each neuron to its k nearest neighbors - equal density everywhere"""
    n = len(coords)
    adjacency = np.zeros((n, n))
    for i in range(n):
        dists = np.linalg.norm(coords - coords[i], axis=1)
        dists[i] = np.inf  # exclude self
        nearest = np.argsort(dists)[:k]
        adjacency[i, nearest] = 1
    return adjacency

def energy_stats(coords, adjacency):
    total_e = 0
    dists = []
    for i in range(len(coords)):
        targets = np.where(adjacency[i] > 0)[0]
        for j in targets:
            d = np.linalg.norm(coords[i] - coords[j])
            total_e += d ** 2
            dists.append(d)
    return total_e, np.mean(dists), np.max(dists)

configs = {
    'grid_2d': lambda tg: tg.grid_2d(32, 32),
    'columnar_3d': lambda tg: tg.columnar_3d(10, 10, 10),
    'loose_clusters (10, r3)': lambda tg: tg.random_cluster(10, cluster_radius=3.0),
    'tight_clusters (5, r1)': lambda tg: tg.random_cluster(5, cluster_radius=1.0),
}

print(f"\nEach neuron connects to its 10 nearest neighbors (10,000 total connections)\n")

results = {}
for name, gen in configs.items():
    tg = TopologyGenerator(1000)
    gen(tg)
    coords = tg.get_coordinates()
    adj = build_k_nearest(coords, k=10)
    total_e, avg_d, max_d = energy_stats(coords, adj)
    results[name] = total_e
    print(f"{name}:")
    print(f"  Total energy: {total_e:,.0f} | Avg wire length: {avg_d:.3f} | Longest wire: {max_d:.2f}")

print("\n" + "=" * 60)
print("FINAL RANKING (same workload, lower = better chip)")
print("=" * 60)
for name in sorted(results, key=results.get):
    baseline = max(results.values())
    savings = (1 - results[name]/baseline) * 100
    print(f"  {name}: {results[name]:,.0f}  ({savings:.0f}% less energy than worst)")
```

```python
import sys
sys.path.insert(0, r'C:\Users\Bo\projects\neuromorphic-sandbox')
from src.topology import TopologyGenerator
import numpy as np

print("=" * 60)
print("EXPERIMENT 4: HEAT-CONSTRAINED PLACEMENT")
print("Finding optimal cluster tightness under thermal limits")
print("=" * 60)

def build_k_nearest(coords, k=10):
    n = len(coords)
    adjacency = np.zeros((n, n))
    for i in range(n):
        dists = np.linalg.norm(coords - coords[i], axis=1)
        dists[i] = np.inf
        nearest = np.argsort(dists)[:k]
        adjacency[i, nearest] = 1
    return adjacency

def comm_energy(coords, adjacency):
    """Communication energy: sum of distance^2"""
    total = 0
    for i in range(len(coords)):
        for j in np.where(adjacency[i] > 0)[0]:
            total += np.linalg.norm(coords[i] - coords[j]) ** 2
    return total

def thermal_penalty(coords, cell_size=1.0, density_limit=8, penalty_scale=50):
    """Divide space into cells. Neurons per cell above the limit
    incur a superlinear (quadratic) penalty - modeling heat buildup."""
    # Bin neurons into grid cells
    mins = coords.min(axis=0)
    cells = {}
    for c in coords:
        key = tuple(((c - mins) // cell_size).astype(int))
        cells[key] = cells.get(key, 0) + 1
    
    penalty = 0
    max_density = 0
    for count in cells.values():
        max_density = max(max_density, count)
        if count > density_limit:
            excess = count - density_limit
            penalty += penalty_scale * excess ** 2  # quadratic: heat compounds
    return penalty, max_density

print("\nSweeping cluster radius from very tight to very loose")
print(f"(5 clusters, 1000 neurons, k=10 connections each)")
print(f"(Thermal limit: 8 neurons per unit cell, quadratic penalty above)\n")

radii = [0.3, 0.5, 0.75, 1.0, 1.5, 2.0, 3.0, 4.0, 6.0]
results = []

for r in radii:
    np.random.seed(42)  # same cluster centers each run - fair comparison
    tg = TopologyGenerator(1000)
    tg.random_cluster(5, cluster_radius=r)
    coords = tg.get_coordinates()
    adj = build_k_nearest(coords, k=10)
    
    e_comm = comm_energy(coords, adj)
    e_heat, max_d = thermal_penalty(coords)
    e_total = e_comm + e_heat
    
    results.append((r, e_comm, e_heat, e_total, max_d))
    print(f"radius={r:4.2f}:  comm={e_comm:10,.0f}  heat={e_heat:12,.0f}  TOTAL={e_total:12,.0f}  (max density: {max_d}/cell)")

print("\n" + "=" * 60)
best = min(results, key=lambda x: x[3])
print(f"OPTIMAL CLUSTER RADIUS: {best[0]}")
print(f"  Communication energy: {best[1]:,.0f}")
print(f"  Thermal penalty:      {best[2]:,.0f}")
print(f"  Total cost:           {best[3]:,.0f}")
print("=" * 60)
print("\nThe curve should be U-shaped:")
print("  Too tight -> thermal penalty explodes")
print("  Too loose -> communication energy explodes")
print("  The optimum is the chip designer's sweet spot.")
```

```python
import sys
sys.path.insert(0, r'C:\Users\Bo\projects\neuromorphic-sandbox')
import numpy as np
from collections import deque

print("EXPERIMENT 5: SMALL-WORLD HYBRID (fast version)")
print("=" * 60)

def spread_clusters(n_neurons, n_clusters, radius, center_spread, seed=42):
    coords, cluster_ids = [], []
    per_cluster = n_neurons // n_clusters
    rng = np.random.default_rng(seed)
    centers = rng.uniform(0, center_spread, size=(n_clusters, 2))
    for cid, (cx, cy) in enumerate(centers):
        for _ in range(per_cluster):
            theta = rng.uniform(0, 2*np.pi)
            r = rng.uniform(0, radius)
            coords.append([cx + r*np.cos(theta), cy + r*np.sin(theta), 0.0])
            cluster_ids.append(cid)
    return np.array(coords), np.array(cluster_ids)

def build_k_nearest(coords, k=10):
    n = len(coords)
    adjacency = np.zeros((n, n))
    for i in range(n):
        dists = np.linalg.norm(coords - coords[i], axis=1)
        dists[i] = np.inf
        adjacency[i, np.argsort(dists)[:k]] = 1
    return adjacency

def comm_energy(coords, adjacency):
    total = 0
    for i in range(len(coords)):
        for j in np.where(adjacency[i] > 0)[0]:
            total += np.linalg.norm(coords[i] - coords[j]) ** 2
    return total

def add_highways(adjacency, cluster_ids, n_highways, coords, seed=1):
    adj = adjacency.copy()
    rng = np.random.default_rng(seed)
    n = len(cluster_ids)
    added, hw_energy = 0, 0
    while added < n_highways:
        i, j = rng.integers(0, n, size=2)
        if cluster_ids[i] != cluster_ids[j] and adj[i, j] == 0:
            adj[i, j] = adj[j, i] = 1
            hw_energy += 2 * np.linalg.norm(coords[i] - coords[j]) ** 2
            added += 1
    return adj, hw_energy

def avg_hops(adjacency, n_samples=30, seed=7):
    n = len(adjacency)
    neighbors = [np.where(adjacency[i] > 0)[0] for i in range(n)]
    rng = np.random.default_rng(seed)
    total_hops, total_pairs, unreachable = 0, 0, 0
    for s in rng.integers(0, n, size=n_samples):
        dist = -np.ones(n, dtype=int); dist[s] = 0
        q = deque([s])
        while q:
            u = q.popleft()
            for v in neighbors[u]:
                if dist[v] < 0:
                    dist[v] = dist[u] + 1
                    q.append(v)
        reached = dist[dist > 0]
        total_hops += reached.sum()
        total_pairs += len(reached)
        unreachable += (dist < 0).sum()
    return total_hops / max(total_pairs, 1), unreachable / n_samples

print("Building baseline (r=8 clusters, k=10)...")
coords, cluster_ids = spread_clusters(1000, 5, radius=8, center_spread=40)
base_adj = build_k_nearest(coords, k=10)
base_energy = comm_energy(coords, base_adj)
base_hops, base_unreach = avg_hops(base_adj)
print(f"Baseline: energy={base_energy:,.0f} | hops={base_hops:.2f} | unreachable={base_unreach:.0f}/source\n")

print(f"{'highways':>9} | {'energy':>10} | {'vs base':>8} | {'hops':>6} | {'unreachable':>11}")
print("-" * 58)
for n_hw in [0, 5, 10, 25, 50, 100]:
    if n_hw == 0:
        adj, hw_e = base_adj, 0
    else:
        adj, hw_e = add_highways(base_adj, cluster_ids, n_hw, coords)
    e = base_energy + hw_e
    hops, unreach = avg_hops(adj)
    print(f"{n_hw:>9} | {e:>10,.0f} | {(e/base_energy-1)*100:>+7.1f}% | {hops:>6.2f} | {unreach:>11.0f}")

print("\nDone!")
```

```python
import sys
sys.path.insert(0, r'C:\Users\Bo\projects\neuromorphic-sandbox')
import numpy as np
from collections import deque

print("EXPERIMENT 6: HUB-ROUTED HIGHWAYS (thalamus / NoC model)")
print("=" * 64)

def spread_clusters(n_neurons, n_clusters, radius, center_spread, seed=42):
    coords, cluster_ids = [], []
    per_cluster = n_neurons // n_clusters
    rng = np.random.default_rng(seed)
    centers = rng.uniform(0, center_spread, size=(n_clusters, 2))
    for cid, (cx, cy) in enumerate(centers):
        for _ in range(per_cluster):
            theta = rng.uniform(0, 2*np.pi)
            r = rng.uniform(0, radius)
            coords.append([cx + r*np.cos(theta), cy + r*np.sin(theta), 0.0])
            cluster_ids.append(cid)
    return np.array(coords), np.array(cluster_ids), centers

def build_k_nearest(coords, k=10):
    n = len(coords)
    adjacency = np.zeros((n, n))
    for i in range(n):
        dists = np.linalg.norm(coords - coords[i], axis=1)
        dists[i] = np.inf
        adjacency[i, np.argsort(dists)[:k]] = 1
    return adjacency

def comm_energy(coords, adjacency):
    total = 0
    for i in range(len(coords)):
        for j in np.where(adjacency[i] > 0)[0]:
            total += np.linalg.norm(coords[i] - coords[j]) ** 2
    return total

def avg_hops(adjacency, n_samples=30, seed=7):
    n = len(adjacency)
    neighbors = [np.where(adjacency[i] > 0)[0] for i in range(n)]
    rng = np.random.default_rng(seed)
    total_hops, total_pairs, unreachable = 0, 0, 0
    for s in rng.integers(0, n, size=n_samples):
        dist = -np.ones(n, dtype=int); dist[s] = 0
        q = deque([s])
        while q:
            u = q.popleft()
            for v in neighbors[u]:
                if dist[v] < 0:
                    dist[v] = dist[u] + 1
                    q.append(v)
        reached = dist[dist > 0]
        total_hops += reached.sum()
        total_pairs += len(reached)
        unreachable += (dist < 0).sum()
    return total_hops / max(total_pairs, 1), unreachable / n_samples

def find_routers(coords, cluster_ids, centers):
    """Router = neuron closest to its cluster's center"""
    routers = []
    for cid in range(len(centers)):
        members = np.where(cluster_ids == cid)[0]
        dists = np.linalg.norm(coords[members, :2] - centers[cid], axis=1)
        routers.append(members[np.argmin(dists)])
    return routers

def add_backbone(adjacency, coords, routers, pattern):
    adj = adjacency.copy()
    hw_energy = 0
    links = []
    R = len(routers)
    if pattern == 'mesh':
        links = [(a, b) for a in range(R) for b in range(a+1, R)]
    elif pattern == 'ring':
        links = [(a, (a+1) % R) for a in range(R)]
    elif pattern == 'star':
        links = [(0, b) for b in range(1, R)]
    for a, b in links:
        i, j = routers[a], routers[b]
        adj[i, j] = adj[j, i] = 1
        hw_energy += 2 * np.linalg.norm(coords[i] - coords[j]) ** 2
    return adj, hw_energy, len(links)

# Build baseline
coords, cluster_ids, centers = spread_clusters(1000, 5, radius=8, center_spread=40)
base_adj = build_k_nearest(coords, k=10)
base_energy = comm_energy(coords, base_adj)
routers = find_routers(coords, cluster_ids, centers)

print(f"\nBaseline local energy: {base_energy:,.0f}")
print(f"Routers elected (node ids): {routers}\n")

print(f"{'topology':>12} | {'links':>5} | {'energy':>10} | {'vs base':>8} | {'hops':>6} | {'unreach':>7}")
print("-" * 66)

# Baseline for reference
hops, unreach = avg_hops(base_adj)
print(f"{'no backbone':>12} | {0:>5} | {base_energy:>10,.0f} | {'+0.0%':>8} | {hops:>6.2f} | {unreach:>7.0f}")

for pattern in ['star', 'ring', 'mesh']:
    adj, hw_e, n_links = add_backbone(base_adj, coords, routers, pattern)
    e = base_energy + hw_e
    hops, unreach = avg_hops(adj)
    print(f"{pattern:>12} | {n_links:>5} | {e:>10,.0f} | {(e/base_energy-1)*100:>+7.1f}% | {hops:>6.2f} | {unreach:>7.0f}")

print("\nComparison targets from Experiment 5 (random highways):")
print("   5 random | energy +44.2% | hops 10.98")
print("  25 random | energy +223%  | hops  7.41")
print("  50 random | energy +460%  | hops  6.40")
```
